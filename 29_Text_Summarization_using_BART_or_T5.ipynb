```python
from transformers import pipeline

# 1. Load Pre-trained GPT2 Model under the unified "text-generation" pipeline
print("Loading GPT-2 Model...")

generator = pipeline(
    "text-generation",
    model="gpt2"
)

# 2. Input Text
text = """
Artificial Intelligence is one of the fastest growing technologies
in the modern world. It enables computers and machines to perform
tasks that normally require human intelligence.

Machine Learning is a branch of Artificial Intelligence that allows
systems to learn from data and improve their performance without
being explicitly programmed.

Deep Learning is a part of Machine Learning that uses neural networks
to solve complex problems such as image recognition, speech processing,
and natural language understanding.

Artificial Intelligence is widely used in healthcare, education,
banking, transportation, and many other industries.
"""

prompt = f"Summarize the following text:\n{text}\nSummary:"

print("\nOriginal Text:")
print(text)

# 3. Generate Summary
outputs = generator(
    prompt,
    max_new_tokens=50,
    do_sample=False,
    pad_token_id=50256
)

# 4. Display Summary
generated_text = outputs[0]["generated_text"]
summary = generated_text[len(prompt):].strip()

print("\nGenerated Summary:")
print(summary if summary else generated_text)

print("\nSummarization Completed Successfully!")
```
